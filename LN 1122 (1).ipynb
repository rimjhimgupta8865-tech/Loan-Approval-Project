```python
df.isnull().sum()
```

```python
df['Gender'].fillna(df['Gender'].mode()[0], inplace=True)
```

```python
df['Gender']
```

```python
df['Married'].fillna(df['Married'].mode()[0], inplace=True)
```

```python
df['Married']
```

```python
df['Self_Employed'].fillna(df['Self_Employed'].mode()[0], inplace=True)
```

```python
df['Self_Employed']
```

```python
df.fillna(
    {'Loan_Amount_Term': df['Loan_Amount_Term'].median()},
    inplace=True
)
```

```python
df['Loan_Amount_Term']
```

```python
df.fillna(
    {'Credit_History': df['Credit_History'].mode()[0]},
    inplace=True
)
```

```python
df['Credit_History']
```

Bullet Points

* Missing values identified and treated.
* Median used for numerical columns.
* Mode used for categorical columns.

# Section 3: Feature Engineering

```python
df['TotalIncome'] 
```

Created TotalIncome feature to represent household earning capacity.

# Section 4: Univariate Analysis

### Graph 1: Applicant Income Histogram

```python
plt.figure(figsize=(10,5))

sns.histplot(
    df['ApplicantIncome'],
    kde=True
)

plt.title('Applicant Income Distribution')
plt.show()
```

Insights

* Income distribution is right-skewed.
* Majority of applicants fall within lower income ranges.
* Few high-income outliers exist.

### Graph 2: Loan Amount Histogram

```python
plt.figure(figsize=(10,5))

sns.histplot(
    df['LoanAmount'],
    kde=True
)

plt.title('Loan Amount Distribution')
plt.show()
```

Insights

* Most loans are concentrated within a moderate range.
* Large loan requests are less frequent.

### Graph 3: Applicant Income Boxplot

```python
plt.figure(figsize=(10,4))

sns.boxplot(
    x=df['ApplicantIncome']
)

plt.title('Applicant Income Outliers')
plt.show()
```

Insights

* Presence of significant outliers.
* Indicates large variation among applicants.

# Section 5: Categorical Analysis

### Graph 4: Gender Distribution

```python
sns.countplot(
    x='Gender',
    data=df
)

plt.title('Gender Distribution')
plt.show()
```

Insights

* Male applicants dominate the dataset.

### Graph 5: Education Distribution

```python
sns.countplot(
    x='Education',
    data=df
)

plt.title('Education Distribution')
plt.show()
```

Insights

* Majority applicants are graduates.

### Graph 6: Property Area Distribution

```python
sns.countplot(
    x='Property_Area',
    data=df
)

plt.title('Property Area Distribution')
plt.show()
```

Insights

* Urban and Semiurban areas contribute the most applicants.

### Graph 7: Pie Chart

```python
df['Education'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title('Education Composition')
plt.ylabel('')
plt.show()
```

Insights

* Shows proportion of graduates vs non-graduates.

# Section 6: Bivariate Analysis

### Graph 8: Total Income vs Loan Amount

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='TotalIncome',
    y='LoanAmount',
    data=df
)

plt.title('Total Income vs Loan Amount')
plt.show()
```

Insights 

* Positive relationship observed.
* Higher income applicants generally request larger loans.

### Graph 9: Education vs Loan Amount

```python
sns.boxplot(
    x='Education',
    y='LoanAmount',
    data=df
)
```

Insights

* Loan distributions vary by education category.

### Graph 10: Property Area vs Loan Amount

```python
sns.boxplot(
    x='Property_Area',
    y='LoanAmount',
    data=df
)
```

Insights

* Loan demand differs across property locations

# Section 7: Multivariate Analysis

```python
sns.pairplot(
    df[
        [
            'ApplicantIncome',
            'CoapplicantIncome',
            'LoanAmount',
            'TotalIncome'
        ]
    ]
)

plt.show()
```

Insights

* Helps identify relationships among multiple variables simultaneously.

Multivariate Analysis (Pairplot)
* Pairplot was used to analyze relationships among ApplicantIncome, CoapplicantIncome, LoanAmount, and TotalIncome.
* A positive relationship was observed between ApplicantIncome and TotalIncome.
* LoanAmount tends to increase with TotalIncome.
* The dataset contains a few outliers with exceptionally high income values.
* Most observations are concentrated in lower income and loan amount ranges.

# Section 8: Correlation Analysis

Graph 12: Heatmap

```python
plt.figure(figsize=(10,6))

corr = df.corr(numeric_only=True)

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Matrix')
plt.show()
```

Insights

* Identifies relationships among numerical variables.
* Useful for understanding variable dependencies.
  
  Correlation Analysis:
* ApplicantIncome and TotalIncome show a strong positive correlation (0.89).
* LoanAmount has a moderate positive correlation with TotalIncome (0.53) and ApplicantIncome (0.49).
* Credit_History has very weak correlation with other numerical variables.
* No strong negative correlations were observed in the dataset.

# Section 9: Stacked Bar Chart

```python
pd.crosstab(
    df['Property_Area'],
    df['Education']
).plot(
    kind='bar',
    stacked=True,
    figsize=(8,5)
)

plt.title('Education by Property Area')
plt.show()
```

Insights

* Compares education levels across property areas.
